### 1.6 Sets: Unique Values and Deduplication

A **set** stores values **without duplicates**. Even if you create a set with duplicate items, Python keeps only one copy of each! Sets use curly braces `{}`.

```text
visited_cities = {"Paris", "Rome", "Paris"}
                  └── "Paris" is only stored once!
```

**Goal:** Create `visited_cities` with a duplicate entry and observe automatic deduplication.

**Expected output:** `Unique cities: {'Paris', 'Rome'}` (order may vary)

````python
# Notice "Paris" is listed twice!
visited_cities = {"Paris", "Rome", "Paris"}

print("Unique cities:", visited_cities)
````

### 1.7 Sets: Adding Items and Testing Membership (`in`)

- `.add(value)` adds a new item to the set.
- `in` checks whether an item exists in the set (returns `True` or `False`).

```text

```

**Goal:** Add `"Tokyo"` to `visited_cities` and test if `"Paris"` is in the set.

**Expected output:**
```text

```

````python
# TODO: Add "Tokyo" to visited_cities using the .add() method


print("Updated cities:", visited_cities)
print("Has Paris?", "Paris" in visited_cities)
````
