# Python File Handling Workshop

Welcome to the File Handling Workshop! In Python programming, data is rarely hard-coded inside scripts. Whether you are logging application events, analyzing experimental data in CSV spreadsheets, or reading configuration from JSON web APIs, mastering **File I/O (Input/Output)** is an essential software skill.

```text
┌────────────────────────────────────────────────────────────────────────┐
│                        Python File I/O Lifecycle                       │
│                                                                        │
│   Disk Storage                    Python Memory                        │
│   ┌────────────┐   open(..., 'r') ┌──────────────────────────────┐     │
│   │ story.txt  │ ───────────────► │ f.read() / for line in f     │     │
│   │ scores.csv │                  │ np.genfromtxt()              │     │
│   │ data.json  │                  │ json.load()                  │     │
│   └────────────┘                  └──────────────┬───────────────┘     │
│         ▲                                        │                     │
│         └──────── open(..., 'w'/'a') ────────────┘                     │
│                   f.write() / json.dump()                              │
└────────────────────────────────────────────────────────────────────────┘
```

## Learning Objectives

By the end of this workshop, you will be able to:
- Open, read, and write plain text files safely using the `with open(...)` context manager.
- Read files entirely at once (`.read()`) vs. line-by-line using a memory-efficient `for` loop.
- Differentiate between write mode (`'w'`), append mode (`'a'`), and read mode (`'r'`).
- Load structured numerical arrays from CSV files directly into NumPy using `np.genfromtxt()`.
- Parse, manipulate, and export hierarchical structured data using Python's `json` module.
- Safely check for file existence and remove files programmatically using `os.path.exists()` and `os.remove()`.
- Debug common file I/O runtime errors such as `io.UnsupportedOperation` and `FileNotFoundError`.

**Goal:** Complete all 7 hands-on file handling tasks and verify each with automated self-check tests.

````python
# RUN THIS CELL FIRST to generate the necessary practice files.
import json
import csv
import os

# 1. Create story.txt
with open('story.txt', 'w', encoding='utf-8') as f:
    f.write("Once upon a time in a land of code.\n")
    f.write("There lived a Python script.\n")
    f.write("It loved to process data every day.\n")

# 2. Create scores.csv
with open('scores.csv', 'w', newline='', encoding='utf-8') as f:
    writer = csv.writer(f)
    writer.writerow([10, 20, 30])
    writer.writerow([15, 25, 35])
    writer.writerow([12, 22, 32])

# 3. Create user_data.json
data = {
    "name": "Alice",
    "role": "Developer",
    "id": 101
}
with open('user_data.json', 'w', encoding='utf-8') as f:
    json.dump(data, f, indent=4)

print("✅ Setup complete! Practice files created: story.txt, scores.csv, user_data.json")
````

### Task 1: Reading Entire File at Once
**Goal:** Open `story.txt` and read the entire content into a single string.

```text
with open('story.txt', 'r', encoding='utf-8') as f:
    content = f.read()
```

**Instructions:**
1. Use the `with open(...)` statement to open `'story.txt'` in read mode (`'r'`) with `encoding='utf-8'`.
2. Use the `.read()` method to read the file's entire content into a variable named `content`.
3. Print `content`.

> [!NOTE]
> The `with` statement creates a **context manager** that guarantees the file handle is automatically closed as soon as the indented block finishes—even if an error occurs.

**Expected output:**
```text
Once upon a time in a land of code.
There lived a Python script.
It loved to process data every day.
```

````python
# TODO 1: Open 'story.txt' in read mode ('r') with encoding='utf-8' using with open(...)
with open('story.txt', 'r', encoding='utf-8') as f:
    # TODO 2: Read the entire content into a variable named content using .read()
    content = f.read()

# TODO 3: Print the content
print(content)

# Verification
assert isinstance(content, str) and len(content) > 0, "content should not be empty!"
assert "Once upon a time" in content, "story.txt content mismatch!"
print("✅ Task 1 passed!")
````

### Task 2: Reading Line-by-Line Efficiently
**Goal:** Process a file one line at a time using a `for` loop.

When files are gigabytes in size, `.read()` would consume all available RAM. Iterating over the file object `for line in f:` streams one line into memory at a time.

```text
for line in f:
    print(line.strip())  # Removes trailing newline character '\n'
```

**Instructions:**
1. Open `'story.txt'` in read mode (`'r'`) with `encoding='utf-8'`.
2. Iterate through `f` using a `for` loop.
3. Clean each line with `.strip()` and print it.

**Expected output:**
```text
Once upon a time in a land of code.
There lived a Python script.
It loved to process data every day.
```

````python
lines_read = []

# TODO 1: Open 'story.txt' in read mode ('r') with encoding='utf-8'
with open('story.txt', 'r', encoding='utf-8') as f:
    # TODO 2: Iterate through each line in f using a for loop
    for line in f:
        # TODO 3: Strip whitespace from line and print it
        clean_line = line.strip()
        print(clean_line)
        lines_read.append(clean_line)

# Verification
assert len(lines_read) == 3, f"Expected 3 lines, got {len(lines_read)}"
assert lines_read[0] == "Once upon a time in a land of code.", "Line 1 mismatch!"
print("✅ Task 2 passed!")
````

### Task 3: Writing and Appending
**Goal:** Create a new text file in write mode (`'w'`), then append additional content to it in append mode (`'a'`).

```text
Mode 'w'  ──►  Overwrites or creates new file (destructive)
Mode 'a'  ──►  Adds content to the end without deleting existing text
```

**Instructions:**
1. Open `'my_notes.txt'` in write mode (`'w'`) and write `"This is my first note."`.
2. Re-open `'my_notes.txt'` in append mode (`'a'`) and write `"\nThis is an appended note."`.
3. Open `'my_notes.txt'` in read mode (`'r'`) and print the entire file to verify.

**Expected output:**
```text
This is my first note.
This is an appended note.
```

````python
# TODO 1: Open 'my_notes.txt' in write mode ('w') and write "This is my first note."
with open('my_notes.txt', 'w', encoding='utf-8') as f:
    f.write("This is my first note.")

# TODO 2: Open 'my_notes.txt' in append mode ('a') and write "\nThis is an appended note."
with open('my_notes.txt', 'a', encoding='utf-8') as f:
    f.write("\nThis is an appended note.")

# TODO 3: Open 'my_notes.txt' in read mode ('r'), read into notes_content, and print
with open('my_notes.txt', 'r', encoding='utf-8') as f:
    notes_content = f.read()

print(notes_content)

# Verification
assert "This is my first note." in notes_content, "First note missing!"
assert "This is an appended note." in notes_content, "Appended note missing!"
print("✅ Task 3 passed!")
````

### Task 4: Numerical Data with NumPy & CSV
**Goal:** Ingest numerical CSV data directly into a NumPy `ndarray` and compute statistical metrics.

```text
CSV File on Disk               NumPy 2D Array in Memory
10, 20, 30        ──►         [[10., 20., 30.],
15, 25, 35   np.genfromtxt()   [15., 25., 35.],
12, 22, 32                     [12., 22., 32.]]
```

**Instructions:**
1. Import `numpy as np`.
2. Use `np.genfromtxt('scores.csv', delimiter=',')` to load the dataset into variable `scores`.
3. Print the array.
4. Calculate and print the overall average of `scores` using `np.mean()`.

**Expected output:**
```text
Scores array:
 [[10. 20. 30.]
 [15. 25. 35.]
 [12. 22. 32.]]
Mean score: 22.33
```

````python
import numpy as np

# TODO 1: Use np.genfromtxt to load 'scores.csv' with delimiter=',' into variable scores
scores = np.genfromtxt('scores.csv', delimiter=',')

print("Scores array:")
print(scores)

# TODO 2: Calculate the mean of scores using np.mean() into variable mean_score
mean_score = np.mean(scores)

print(f"Mean score: {mean_score:.2f}")

# Verification
assert isinstance(scores, np.ndarray), "scores must be a NumPy array!"
assert scores.shape == (3, 3), f"Expected shape (3, 3), got {scores.shape}"
assert np.isclose(mean_score, 22.333, atol=1e-2), f"Expected ~22.33, got {mean_score}"
print("✅ Task 4 passed!")
````

### Task 5: JSON Round-Trip (Load, Modify, and Dump)
**Goal:** Parse structured JSON data, update an attribute in Python memory, and write the updated data back to disk.

```text
Disk File (user_data.json)  ──► json.load(f)  ──► Python Dict
Python Dict (modified)       ──► json.dump(...) ──► Disk File (user_data_updated.json)
```

**Instructions:**
1. Import the `json` library.
2. Open `'user_data.json'` in read mode (`'r'`) and parse it into a Python dictionary named `user_data` using `json.load(f)`.
3. Update the value of the key `'role'` to `'Data Scientist'`.
4. Open `'user_data_updated.json'` in write mode (`'w'`) and save the modified dictionary using `json.dump(user_data, f, indent=4)`.
5. Print `user_data`.

**Expected output:**
```text
Modified user data: {'name': 'Alice', 'role': 'Data Scientist', 'id': 101}
```

````python
import json
import os

# TODO 1: Open 'user_data.json' in read mode and parse with json.load()
with open('user_data.json', 'r', encoding='utf-8') as f:
    user_data = json.load(f)

# TODO 2: Update the 'role' key to 'Data Scientist'
user_data['role'] = 'Data Scientist'

# TODO 3: Open 'user_data_updated.json' in write mode ('w') and write with json.dump(..., indent=4)
with open('user_data_updated.json', 'w', encoding='utf-8') as f:
    json.dump(user_data, f, indent=4)

print("Modified user data:", user_data)

# Verification
assert user_data['role'] == 'Data Scientist', "Role was not updated to Data Scientist!"
assert os.path.exists('user_data_updated.json'), "user_data_updated.json file was not created!"
print("✅ Task 5 passed!")
````

### Task 6: Defensive File Cleanup with `os`
**Goal:** Verify whether a file exists before attempting to delete it.

Calling `os.remove()` on a non-existent file triggers a fatal `FileNotFoundError`. Defensive programming checks `os.path.exists()` first.

```text
if os.path.exists('my_notes.txt'):
    os.remove('my_notes.txt')
```

**Instructions:**
1. Import `os`.
2. Check if `'my_notes.txt'` exists using `os.path.exists()`.
3. If it exists, delete it using `os.remove('my_notes.txt')` and print a success message.
4. If it does not exist, print that the file was not found.

**Expected output:**
```text
'my_notes.txt' was successfully deleted.
```

````python
import os

target_file = 'my_notes.txt'

# TODO 1: Check if target_file exists with os.path.exists()
if os.path.exists(target_file):
    # TODO 2: Delete target_file using os.remove()
    os.remove(target_file)
    print(f"'{target_file}' was successfully deleted.")
else:
    print(f"'{target_file}' was not found.")

# Verification
assert not os.path.exists(target_file), f"'{target_file}' should have been deleted!"
print("✅ Task 6 passed!")
````

### Task 7: The Bug Hunt (Traceback Debugging)
**Goal:** Identify and fix a common Python file I/O runtime error.

A frequent beginner mistake is opening a file in read mode (`'r'`) and attempting to write to it, which triggers an `io.UnsupportedOperation: not writable` error.

Run the cell below as-is to inspect the deliberate traceback:

```text
UnsupportedOperation: not writable
```

**Task:**
1. Observe the error message in the traceback.
2. Fix the file opening mode so that writing a new log message to `'practice_log.txt'` succeeds.

**Expected output:**
```text
Log content:
File handling mastered!
```

````python
# BUG HUNT: The code below fails because the file was opened in read mode ('r').
# Fix the mode on the open() call so writing succeeds!

# Fixed mode: open in write mode ('w') so write() succeeds
with open('practice_log.txt', 'w', encoding='utf-8') as f:
    f.write("File handling mastered!\n")

with open('practice_log.txt', 'r', encoding='utf-8') as f:
    log_content = f.read()

print("Log content:")
print(log_content.strip())

# Verification
assert "File handling mastered!" in log_content, "Log entry missing!"
print("✅ Bug successfully identified and fixed!")
````
