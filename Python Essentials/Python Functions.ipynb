### 1.2 Multiple arguments: order matters

A function can accept several inputs. Each input lands in a labeled slot called a parameter. The order of the call decides which value goes into which slot.

```text

                    │  │  │
                    │  │  └─ third slot
                    │  └─ second slot
                    └─ first slot
```

### 1.6 The `map` function: an assembly line

`map` applies a function to every item of a list, one item at a time, like an assembly line feeding each part through the same machine. It hands back a lazy sequence, so you wrap it in `list(...)` to collect and print the results.

```text
list( map(times_ten, numbers) )
 │     │    │           │
 │     │    │           └─ the list whose items go down the line
 │     │    └─ the function applied to every item
 │     └─ built-in map: builds the results lazily
 └─ list(...) pours the lazy results into a real list

list(map(times_ten, [1, 2, 3]))  →  [10, 20, 30]
```

- `map` takes two inputs: the function first, then the list.
- The results come back lazily; `list(...)` materializes them.
