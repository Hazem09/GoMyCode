## Part 1 — Guided Mechanics

### 1.1 Creating a Series from a List

A **Series** is a one-dimensional labeled array in Pandas. It pairs each data point with an explicit or default integer **index**.

```text
names = ["Alice", "Bob", "Carlos"]
               │
          pd.Series(...)
               ▼
Index    Value
  0      Alice
  1      Bob
  2      Carlos
dtype: object
```

**Goal:** Convert the Python list `["Alice", "Bob", "Carlos"]` into a Pandas Series named `series`. Display the Series and print its `.index` attribute.

**Expected output:**
```text
0     Alice
1       Bob
2    Carlos
dtype: object
RangeIndex(start=0, stop=3, step=1)
```

````python
names = ["Alice", "Bob", "Carlos"]

series = pd.Series(names)

print(series)
print(series.index)
````

### 1.2 Creating a DataFrame from a Dictionary

A **DataFrame** is a 2D table composed of rows and columns. When creating a DataFrame from a Python dictionary, the dictionary **keys** become the column headers, and the **values** (lists) become the column data.

```text
Dictionary: {"Name": ["Ali", "Mira", "Ziad"], "Score": [88, 92, 79]}
                                 │
                          pd.DataFrame(...)
                                 ▼
        Name  Score
0        Ali     88
1       Mira     92
2       Ziad     79
```

**Goal:** Build a DataFrame named `df` from the dictionary `{"Name": ["Ali", "Mira", "Ziad"], "Score": [88, 92, 79]}` and print the resulting table.

**Expected output:**
```text
   Name  Score
0   Ali     88
1  Mira     92
2  Ziad     79
```

````python
data = {
    "Name": ["Ali", "Mira", "Ziad"],
    "Score": [88, 92, 79]
}

df = pd.DataFrame(data)

print(df)
````

### 1.3 Selecting a Single Column

To extract a single column from a DataFrame, use square brackets with the column name string: `df["ColumnName"]`. Selecting a single column yields a 1D Pandas **Series**.

```text
df["Score"]
    ▼
0    88
1    92
2    79
Name: Score, dtype: int64  (type: Series)
```

**Goal:** Select the `"Score"` column from `df`, assign it to `scores`, print it, and verify its type with `type()`.

**Expected output:**
```text
Scores column:
0    88
1    92
2    79
Name: Score, dtype: int64
Type: <class 'pandas.core.series.Series'>
```

````python
scores = df["Score"]

print("Scores column:\n", scores)
print("Type:", type(scores))
````

### 1.4 Selecting Multiple Columns

When you want two or more columns, pass a **list of column names** inside square brackets: `df[["ColA", "ColB"]]`. Notice the double square brackets:
- The outer brackets indicate DataFrame column indexing.
- The inner brackets create the Python list of column names.

Selecting multiple columns always returns a **DataFrame**.

```text
df[["Name", "Score"]]
         ▼
   Name  Score
0   Ali     88
1  Mira     92
2  Ziad     79
```

**Goal:** Select both `"Name"` and `"Score"` columns together, assign the result to `name_score`, and print it.

**Expected output:**
```text
   Name  Score
0   Ali     88
1  Mira     92
2  Ziad     79
```

````python
name_score = df[["Name", "Score"]]

print(name_score)
````

### 1.5 Setting Custom Index and Selecting Rows by Label with `.loc`

By default, DataFrame rows are indexed with integers `0, 1, 2, ...`. You can customize the row index by assigning a list of labels to `df.index`.

Once labeled, use `.loc[label]` to retrieve a row by its exact label:

```text
df.index = ['student_1', 'student_2', 'student_3']

            Name  Score
student_1    Ali     88
student_2   Mira     92  ← df.loc['student_2']
student_3   Ziad     79
```

**Goal:**
1. Assign `['student_1', 'student_2', 'student_3']` to `df.index`.
2. Select the row labeled `'student_2'` using `.loc` and assign it to `student_two`.

**Expected output:**
```text
Name     Mira
Score      92
Name: student_2, dtype: object
```

````python
df.index = ['student_1', 'student_2', 'student_3']

student_two = df.loc['student_2']

print(student_two)
````

### 1.6 Selecting Rows and Columns by Label with `.loc`

The `.loc` accessor supports 2D selection using the syntax:
`df.loc[row_labels, column_labels]`

You can pass lists for both rows and columns to select specific cells by name:

```text
df.loc[['student_1', 'student_3'], ['Score', 'Name']]
                       │
                       ▼
           Score   Name
student_1     88    Ali
student_3     79   Ziad
```

**Goal:** Using the labeled DataFrame from Section 1.5, select the `"Score"` and `"Name"` columns for rows `'student_1'` and `'student_3'`.

**Expected output:**
```text
           Score  Name
student_1     88   Ali
student_3     79  Ziad
```

````python
subset_loc = df.loc[['student_1', 'student_3'], ['Score', 'Name']]

print(subset_loc)
````

### 1.7 Selecting Rows by Integer Position with `.iloc`

While `.loc` searches by **label**, `.iloc` searches strictly by **integer position** (0-based index), regardless of what custom labels exist on the DataFrame.

```text
Position:      Label:       Name  Score
  0         student_1        Ali     88
  1         student_2       Mira     92  ← df.iloc[1] selects position 1
  2         student_3       Ziad     79
```

> [!NOTE]
> Even though your DataFrame has string labels like `'student_2'`, `df.iloc[1]` always accesses the second row because positional counting starts at `0`.

**Goal:** Select the second row (integer position `1`) from `df` using `.iloc` and assign it to `second_row`.

**Expected output:**
```text
Name     Mira
Score      92
Name: student_2, dtype: object
```

````python
second_row = df.iloc[1]

print(second_row)
````

### 1.8 Selecting Rows and Columns by Position with `.iloc`

Just like `.loc`, `.iloc` accepts two coordinates:
`df.iloc[row_positions, col_positions]`

Both row and column coordinates must be integers, slices (`start:stop`), or lists of integers:

```text
Position:      col 0     col 1
               Name      Score
row 0           Ali         88  ← selected
row 1          Mira         92
row 2          Ziad         79  ← selected
                             ▲
                             selected (col 1)
```

**Goal:** Using `.iloc`, select rows at positions `0` and `2`, and the column at position `1` (`"Score"`).

**Expected output:**
```text
student_1    88
student_3    79
Name: Score, dtype: int64
```

````python
subset_iloc = df.iloc[[0, 2], 1]

print(subset_iloc)
````

### 1.9 Adding Columns and Conditional Filtering

You can create a new column in a DataFrame by assigning values directly: `df["NewCol"] = values`.
You can also filter rows using boolean conditions: `df[condition]`.

```text
df["Passed"] = df["Score"] >= 80
df[df["Passed"]]  →  returns only rows where Passed is True
```

**Goal:**
1. Add a new boolean column `"Passed"` that evaluates to `True` if `Score >= 80`, otherwise `False`.
2. Filter and display only the students who passed.

**Expected output:**
```text
Updated DataFrame:
            Name  Score  Passed
student_1    Ali     88    True
student_2   Mira     92    True
student_3   Ziad     79   False

Passing students:
            Name  Score  Passed
student_1    Ali     88    True
student_2   Mira     92    True
```

````python
df["Passed"] = df["Score"] >= 80

passing_students = df[df["Passed"]]

print("Updated DataFrame:\n", df)
print("\nPassing students:\n", passing_students)
````

### 2.1 Product Inventory DataFrame

Create an inventory DataFrame named `inventory` from a dictionary containing 4 items:
- `"Product"`: `["Laptop", "Mouse", "Keyboard", "Monitor"]`
- `"Category"`: `["Electronics", "Accessories", "Accessories", "Electronics"]`
- `"Price"`: `[999.99, 25.50, 49.00, 199.99]`
- `"Stock"`: `[12, 120, 75, 30]`

**Goal:** Create `inventory`, display the table, and print its `.shape` (rows, columns).

**Expected output:**
```text
    Product     Category   Price  Stock
0    Laptop  Electronics  999.99     12
1     Mouse  Accessories   25.50    120
2  Keyboard  Accessories   49.00     75
3   Monitor  Electronics  199.99     30
Shape: (4, 4)
```

````python
inventory_data = {
    "Product": ["Laptop", "Mouse", "Keyboard", "Monitor"],
    "Category": ["Electronics", "Accessories", "Accessories", "Electronics"],
    "Price": [999.99, 25.50, 49.00, 199.99],
    "Stock": [12, 120, 75, 30]
}

inventory = pd.DataFrame(inventory_data)

print(inventory)
print("Shape:", inventory.shape)
````

### 2.2 Price Analysis and Column Extraction

From the `inventory` DataFrame:
1. Extract the `"Product"` and `"Price"` columns into a new DataFrame `product_prices`.
2. Calculate the average price of products (`.mean()`) and identify the maximum price (`.max()`).

**Goal:** Extract multiple columns and compute summary metrics on price.

**Expected output:**
```text
Product and Price:
    Product   Price
0    Laptop  999.99
1     Mouse   25.50
2  Keyboard   49.00
3   Monitor  199.99

Average price: $318.62
Maximum price: $999.99
```

````python
product_prices = inventory[["Product", "Price"]]

avg_price = inventory["Price"].mean()
max_price = inventory["Price"].max()

print("Product and Price:\n", product_prices)
print(f"\nAverage price: ${avg_price:.2f}")
print(f"Maximum price: ${max_price:.2f}")
````

### 2.3 Product SKU Lookup with `.loc`

Assign unique SKU IDs `['SKU-01', 'SKU-02', 'SKU-03', 'SKU-04']` as the index of `inventory`. Then retrieve the `"Product"` and `"Stock"` information for items `'SKU-02'` and `'SKU-04'` using `.loc`.

**Goal:** Re-index `inventory` with custom SKU codes and perform multi-row, multi-column retrieval with `.loc`.

**Expected output:**
```text
         Product  Stock
SKU-02     Mouse    120
SKU-04   Monitor     30
```

````python
inventory.index = ['SKU-01', 'SKU-02', 'SKU-03', 'SKU-04']

sku_subset = inventory.loc[['SKU-02', 'SKU-04'], ['Product', 'Stock']]

print(sku_subset)
````

### 2.4 Fast Positional Slicing with `.iloc`

Using `.iloc`, slice the first two rows (positions `0:2`) and the first three columns (positions `0:3`) of `inventory`.

> [!TIP]
> Just like Python list slicing, `.iloc[0:2, 0:3]` is exclusive of the stop index!

**Goal:** Perform 2D range slicing using integer positions.

**Expected output:**
```text
        Product     Category   Price
SKU-01   Laptop  Electronics  999.99
SKU-02    Mouse  Accessories   25.50
```

````python
top_products = inventory.iloc[0:2, 0:3]

print(top_products)
````

### 2.5 Low Stock & Category Filter

Find products that belong to the `"Accessories"` category and have a `"Stock"` greater than 50.

**Goal:** Apply compound boolean filtering to `inventory` using `&` (and) between conditions.

**Expected output:**
```text
Filtered accessories:
         Product     Category  Price  Stock
SKU-02     Mouse  Accessories   25.5    120
SKU-03  Keyboard  Accessories   49.0     75
```

````python
filtered_items = inventory[(inventory["Category"] == "Accessories") & (inventory["Stock"] > 50)]

print("Filtered accessories:\n", filtered_items)
````
