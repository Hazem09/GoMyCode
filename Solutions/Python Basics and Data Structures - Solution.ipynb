### 1.1 List Slicing: Extracting Sublists

When you need a portion of a list instead of just a single element, use **slicing**. Slicing uses square brackets with a colon: `list[start:stop]`.

- `start` is the index where the slice begins (**inclusive**).
- `stop` is the index where the slice ends (**exclusive** — Python stops right before this index).

```text
scores = [95, 88, 72, 64, 50, 41]
index:     0   1   2   3   4   5
          ├───┴───┤
scores[0:3] → [95, 88, 72]   (indices 0, 1, and 2; stops BEFORE 3)
```

Useful slicing shortcuts:
- `[:3]` starts at index 0 and goes up to index 3 (exclusive).
- `[3:]` starts at index 3 and takes everything through the end of the list.
- `[-1]` accesses the very last element directly.

**Goal:** From `temperatures = [18, 21, 24, 26, 22, 19, 15]`, extract the first 3 days (`first_three`), the weekend days from index 5 onward (`weekend`), and the last temperature (`last_temp`).

**Expected output:**
```text
First three days: [18, 21, 24]
Weekend days: [19, 15]
Last temperature: 15
```

````python
temperatures = [18, 21, 24, 26, 22, 19, 15]

first_three = temperatures[:3]
weekend = temperatures[5:]
last_temp = temperatures[-1]

print("First three days:", first_three)
print("Weekend days:", weekend)
print("Last temperature:", last_temp)
````

### 1.2 Removing Elements: `.remove()` vs `.pop()`

Python provides two primary ways to remove an item from a list:

```text
cart = ["apple", "bread", "milk", "bread"]

cart.remove("bread")   Searches by VALUE: removes the FIRST occurrence of "bread"
                       Result: ["apple", "milk", "bread"]

removed = cart.pop(1)  Removes by INDEX: removes item at position 1 AND returns it
                       Result: cart is ["apple", "bread"], removed is "milk"
```

Key differences:
- `.remove(val)` looks for the *matching value* and deletes it. It does not return the deleted item.
- `.pop(index)` removes the item at that *specific index* and returns it. If you do not pass an index, `.pop()` removes and returns the *last* item.

**Goal:** Given `tasks = ["email", "groceries", "code", "gym"]`, remove `"groceries"` using `.remove()`, and remove the last task using `.pop()` into a variable named `done_task`.

**Expected output:**
```text
Removed task: gym
Remaining tasks: ['email', 'code']
```

````python
tasks = ["email", "groceries", "code", "gym"]

tasks.remove("groceries")
done_task = tasks.pop()

print("Removed task:", done_task)
print("Remaining tasks:", tasks)
````

### 1.3 Conditions with Data Structures: Checking Membership (`in` / `not in`)

You can test whether a value exists inside a list, tuple, set, or dictionary key using the `in` and `not in` operators inside an `if` statement.

```text
guests = ["Alice", "Bob", "Charlie"]

if "Bob" in guests:
    print("Welcome Bob!")
```

Mental model:
- `x in collection` evaluates to `True` if `x` is found, and `False` otherwise.
- For dictionaries, `x in my_dict` checks whether `x` is one of the **keys**.

**Goal:** Check if `"milk"` is in `grocery_list = ["bread", "eggs", "apples"]`. If it is in the list, print `"Milk is already on the list"`. Otherwise, print `"Milk is missing! Adding milk..."` and append `"milk"` to the list.

**Expected output:**
```text
Milk is missing! Adding milk...
Updated groceries: ['bread', 'eggs', 'apples', 'milk']
```

````python
grocery_list = ["bread", "eggs", "apples"]

if "milk" in grocery_list:
    print("Milk is already on the list")
else:
    print("Milk is missing! Adding milk...")
    grocery_list.append("milk")

print("Updated groceries:", grocery_list)
````

### 1.4 Checking List Length and Emptiness (`len()`)

The built-in `len()` function counts how many items are inside a data structure. You can use it in conditional checks to see if a collection has items or is empty.

```text
items = ["pencil", "eraser"]
len(items) → 2

if len(items) > 0:
    print("You have supplies.")
```

> [!TIP]
> In Python, an empty collection like `[]` or `{}` is considered "falsy". Checking `if len(notifications) > 0:` is explicit and beginner-friendly!

**Goal:** Check the length of `notifications = ["New message", "System update"]`. If `len(notifications) > 0`, print `"You have 2 new notifications."` (using `len(notifications)`). Otherwise, print `"No new notifications."`.

**Expected output:** `You have 2 new notifications.`

````python
notifications = ["New message", "System update"]

if len(notifications) > 0:
    print(f"You have {len(notifications)} new notifications.")
else:
    print("No new notifications.")
````

### 1.5 Looping Over a List: The `for` Loop

A `for` loop lets you inspect or process every item in a list from left to right, one by one.

```text
cities = ["Tokyo", "Paris", "Cairo"]

for city in cities:
     │        │
     │        └─ the collection you are looping through
     └─ temporary variable storing current item each round

Round 1: city = "Tokyo"
Round 2: city = "Paris"
Round 3: city = "Cairo"
```

- In each turn of the loop, Python assigns the next item in `cities` to the loop variable `city`.
- The indented block executes once for each element.

**Goal:** Loop through `prices = [12, 45, 9, 30]` and print each price formatted with a dollar sign (e.g. `Price: $12`).

**Expected output:**
```text
Price: $12
Price: $45
Price: $9
Price: $30
```

````python
prices = [12, 45, 9, 30]

for price in prices:
    print(f"Price: ${price}")
````

### 1.6 Looping Over Dictionaries: Keys, Values, and `.items()`

When looping over a dictionary, Python gives you three options:

1. `for key in my_dict:` loops through keys only.
2. `for val in my_dict.values():` loops through values only.
3. `for key, val in my_dict.items():` loops through **both** keys and values together!

```text
fruit_stock = {"apple": 10, "banana": 5, "orange": 8}

for fruit, count in fruit_stock.items():
      │       │              │
      │       │              └─ produces pairs of (key, value)
      │       └─ variable for value (10, 5, 8)
      └─ variable for key ("apple", "banana", "orange")
```

Unpacking with `.items()` is the cleanest and most common way to display or work with dictionary records.

**Goal:** Loop through `capital_cities = {"France": "Paris", "Japan": "Tokyo", "Egypt": "Cairo"}` using `.items()` and print `"The capital of France is Paris."`, etc.

**Expected output:**
```text
The capital of France is Paris.
The capital of Japan is Tokyo.
The capital of Egypt is Cairo.
```

````python
capital_cities = {"France": "Paris", "Japan": "Tokyo", "Egypt": "Cairo"}

for country, capital in capital_cities.items():
    print(f"The capital of {country} is {capital}.")
````

### 1.7 Combining Loops and Conditions: Filtering into a New List

A fundamental data-processing pattern in programming is **filtering**:
1. Start with an empty list to hold your results: `filtered_list = []`.
2. Loop over your source collection.
3. Use an `if` statement to test each item against a condition.
4. If the item passes the check, `.append()` it to your result list!

```text
scores = [45, 82, 67, 39, 91]
passing = []

for s in scores:
    if s >= 60:
        passing.append(s)
```

Visual flow:
```text
Input: [45, 82, 67, 39, 91]
         ↓   ↓   ↓   ↓   ↓
       <60 ≥60 ≥60 <60 ≥60
         │   │   │   │   │
        skip │   │  skip │
             └───┴───────┴─→ Result: [82, 67, 91]
```

**Goal:** Given `temperatures = [15, 26, 32, 19, 28, 35, 21]`, filter all temperatures that are `> 25` into a new list named `hot_days`.

**Expected output:** `Hot days: [26, 32, 28, 35]`

````python
temperatures = [15, 26, 32, 19, 28, 35, 21]
hot_days = []

for temp in temperatures:
    if temp > 25:
        hot_days.append(temp)

print("Hot days:", hot_days)
````

### 1.8 Combining Loops and Conditions: The Accumulator Pattern

An **accumulator** is a variable that gathers information across loop iterations:
- A **running total** starts at `0` and adds numeric values (`total += value`).
- A **counter** starts at `0` and adds `1` whenever a condition is met (`count += 1`).

```text
numbers = [5, 12, 3, 18, 7]
count_large = 0

for n in numbers:
    if n > 10:
        count_large += 1   # counts how many numbers exceed 10
```

> [!IMPORTANT]
> Notice where the accumulator is defined: **before the loop starts**! If you place `total = 0` inside the loop, it resets back to 0 on every turn.

**Goal:** Given `order_amounts = [25, 60, 15, 120, 45, 80]`:
1. Calculate `total_sales` (sum of all orders).
2. Count `large_orders` (orders that are `$50` or higher).

**Expected output:**
```text
Total sales: $345
Number of large orders: 3
```

````python
order_amounts = [25, 60, 15, 120, 45, 80]

total_sales = 0
large_orders = 0

for order in order_amounts:
    total_sales += order
    if order >= 50:
        large_orders += 1

print("Total sales: $", total_sales, sep="")
print("Number of large orders:", large_orders)
````

### 1.9 Processing a Collection with `while` and `.pop()`

While `for` loops are great when reading a fixed list, a `while` loop is useful when you want to **consume or drain** a collection item by item until nothing remains.

```text
queue = ["Ticket 1", "Ticket 2", "Ticket 3"]

while len(queue) > 0:
    ticket = queue.pop(0)   # remove the first item
    print("Processing", ticket)

# After the loop finishes, queue is empty []
```

Mental model:
- The condition `len(queue) > 0` checks if there are still items left.
- Each call to `.pop(0)` removes one item from the front and shrinks the list.
- When the list is empty, `len(queue)` equals `0`, the condition becomes `False`, and the loop stops.

**Goal:** Use a `while` loop to process all items in `download_queue = ["file1.zip", "file2.pdf", "file3.png"]` using `.pop(0)`, printing `f"Downloading: {current_file}"` for each.

**Expected output:**
```text
Downloading: file1.zip
Downloading: file2.pdf
Downloading: file3.png
All downloads complete! Remaining queue: []
```

````python
download_queue = ["file1.zip", "file2.pdf", "file3.png"]

while len(download_queue) > 0:
    current_file = download_queue.pop(0)
    print(f"Downloading: {current_file}")

print("All downloads complete! Remaining queue:", download_queue)
````

### 2.1 Playlist Management (Slicing and Removing)

You manage a playlist of songs:
`playlist = ["Bohemian Rhapsody", "Hotel California", "Stairway to Heaven", "Sweet Child O' Mine", "Shape of You", "Bad Romance"]`

Perform the following operations:
1. Extract the first 3 songs into a new list named `top_tracks` using slicing.
2. Remove `"Bad Romance"` from `playlist` using `.remove()`.
3. Remove and save the last song from `playlist` into `last_track` using `.pop()`.

**Goal:** Slice, remove, and pop items from an existing list.

**Expected output:**
```text
Top tracks: ['Bohemian Rhapsody', 'Hotel California', 'Stairway to Heaven']
Last track removed: Shape of You
Remaining playlist: ['Bohemian Rhapsody', 'Hotel California', 'Stairway to Heaven', "Sweet Child O' Mine"]
```

### 2.2 Inventory Stock Check (Dictionary with `if / else`)

A computer store tracks inventory counts in a dictionary:
`inventory = {"laptop": 5, "mouse": 0, "keyboard": 12, "monitor": 3}`

Write an `if / else` block to check the availability of `requested_item = "mouse"`:
- First, check if `requested_item in inventory`:
  - If its quantity `inventory[requested_item] > 0`: print `f"{requested_item.title()} is in stock ({inventory[requested_item]} available)."`
  - Otherwise (`else`): print `f"{requested_item.title()} is currently out of stock."`
- If the item is not in the dictionary at all (`else`): print `f"{requested_item.title()} is not sold here."`

**Goal:** Combine dictionary membership testing with conditional value inspection.

**Expected output:** `Mouse is currently out of stock.`

````python
inventory = {"laptop": 5, "mouse": 0, "keyboard": 12, "monitor": 3}
requested_item = "mouse"

if requested_item in inventory:
    if inventory[requested_item] > 0:
        print(f"{requested_item.title()} is in stock ({inventory[requested_item]} available).")
    else:
        print(f"{requested_item.title()} is currently out of stock.")
else:
    print(f"{requested_item.title()} is not sold here.")
````

### 2.3 Student Grade Categorizer (Loop + `if / elif / else` with Dictionary)

You have a dictionary containing student exam scores:
`grades = {"Amira": 92, "Zack": 48, "Laila": 78, "Tariq": 85, "Maya": 59}`

Loop through `grades.items()` and categorize each student based on their score:
- Score `>= 85`: print `f"{name}: Distinction ({score})"`
- Score `>= 50`: print `f"{name}: Pass ({score})"`
- Otherwise (`else`): print `f"{name}: Retake required ({score})"`

**Goal:** Iterate through key-value pairs and apply multi-branch decision logic (`if / elif / else`).

**Expected output:**
```text
Amira: Distinction (92)
Zack: Retake required (48)
Laila: Pass (78)
Tariq: Distinction (85)
Maya: Pass (59)
```

````python
grades = {"Amira": 92, "Zack": 48, "Laila": 78, "Tariq": 85, "Maya": 59}

for name, score in grades.items():
    if score >= 85:
        print(f"{name}: Distinction ({score})")
    elif score >= 50:
        print(f"{name}: Pass ({score})")
    else:
        print(f"{name}: Retake required ({score})")
````

### 2.4 Filtering and Calculating (List Loop with Condition and Accumulators)

You have a list of order values in dollars:
`order_prices = [15, 80, 45, 120, 30, 95, 25, 110]`

Using a single `for` loop over `order_prices`:
1. Filter all orders that cost more than `$50` into a new list named `premium_orders`.
2. Compute the `total_revenue` of all orders combined.
3. Count how many total orders were processed using `order_count`.

**Goal:** Combine list filtering with accumulator variables.

**Expected output:**
```text
Premium orders: [80, 120, 95, 110]
Total revenue: $520
Total orders: 8
```

````python
order_prices = [15, 80, 45, 120, 30, 95, 25, 110]

premium_orders = []
total_revenue = 0
order_count = 0

for price in order_prices:
    total_revenue += price
    order_count += 1
    if price > 50:
        premium_orders.append(price)

print("Premium orders:", premium_orders)
print(f"Total revenue: ${total_revenue}")
print("Total orders:", order_count)
````

### 2.5 Customer Service Queue (`while` Loop with `.pop(0)`)

A helpdesk has a queue of support tickets waiting in order:
`support_queue = ["Ticket #101", "Ticket #102", "Ticket #103", "Ticket #104"]`

Use a `while` loop to simulate processing each ticket in First-In, First-Out (FIFO) order:
- Run the loop while `support_queue` is not empty (`len(support_queue) > 0`).
- Remove the front ticket using `.pop(0)` and assign it to `current_ticket`.
- Print `f"Serving {current_ticket}... Resolved!"`
- After the loop finishes, print `f"Queue status: {len(support_queue)} tickets remaining."`

**Goal:** Process and empty a collection sequentially with a `while` loop and `.pop(0)`.

**Expected output:**
```text
Serving Ticket #101... Resolved!
Serving Ticket #102... Resolved!
Serving Ticket #103... Resolved!
Serving Ticket #104... Resolved!
Queue status: 0 tickets remaining.
```

````python
support_queue = ["Ticket #101", "Ticket #102", "Ticket #103", "Ticket #104"]

while len(support_queue) > 0:
    current_ticket = support_queue.pop(0)
    print(f"Serving {current_ticket}... Resolved!")

print(f"Queue status: {len(support_queue)} tickets remaining.")
````
